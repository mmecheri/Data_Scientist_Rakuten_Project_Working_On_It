## 1. Introduction

This project focuses on image-based product classification using **Transfer Learning**.  
Given the dataset's complexity and visual similarity between classes, we rely on pre-trained CNNs (e.g., DenseNet121, Xception, etc.) to build robust models with limited compute resources.

## 3. Modeling Strategy Overview

We structured the modeling phase into five manual experimentation steps to improve classification performance:

### Step 1 — Baseline Evaluation
- Train selected models without data augmentation or fine-tuning.
- Use default LR (0.001), batch sizes of 32 and 64, and epochs = 10, 20, or 40.
- Select top 5 models based on weighted F1-score.

### Step 2 — Data Augmentation
- Apply standard image augmentation: rotation, shear, zoom, shift, flip, etc.

### Step 3 — Fine-Tuning
- Unfreeze selected layers of top-performing models.
- Train with low LR (0.00001) and longer training (40–60 epochs).

### Step 4 — Learning Rate Optimization
- Use `LearningRateScheduler` to increase LR from 0.00001 to ~2.54.
- Plot loss and accuracy to identify optimal LR.

### Step 5 — Final Training
- Combine optimized LR, fine-tuning, and augmentation for final training.

## 4. Step 1 — Baseline Model Training

### Objective

Establish a baseline performance for various pre-trained CNN models by training them on raw, resized images without any data augmentation or fine-tuning.  
This step aims to identify the most promising architectures under default training conditions and standardized input configurations.

---

### Parameters
- Learning Rate: 0.001
- Epochs: 40 (with EarlyStopping)
- Batch Size: 64

> A batch size of 64 was used in all final runs, as it provided stable training and efficient GPU memory usage.  
Preliminary tests with batch size 32 showed no significant benefit and were not retained for further experiments.


---

### Classification Head Architecture

The following classification layers were added on top of each pre-trained base model (CNN backbone).  
This head is used to project the extracted features into the 27 target classes of the Rakuten dataset.

```python
# Classification Head
model.add(GlobalAveragePooling2D())       # Reduce feature maps to 1D
model.add(Dense(units=1024, activation='relu'))
model.add(Dropout(rate=0.2))
model.add(Dense(units=512, activation='relu'))
model.add(Dropout(rate=0.2))
model.add(Dense(units=27, activation='softmax'))  # Output layer for multi-class classification
```
---

### Data Loading Strategy

Images were loaded using `ImageDataGenerator` with pixel rescaling (1./255) and an 80/20 train-validation split.  
Data was read from a Pandas DataFrame using `flow_from_dataframe`, referencing image file names and class labels.

Labels were encoded as integers (`sparse` mode), and shuffling was applied only to the training set.

A total of **67,933 images** were used for training and **16,983 for validation**, ensuring consistency across all experiments.

### Understanding Input Shape Definitions

| **Concept**              | **Definition**                                                               | **Example**                                  |
|--------------------------|-------------------------------------------------------------------------------|----------------------------------------------|
| **Original Input Shape** | Input size the pre-trained model was trained on (ImageNet).                 | DenseNet121 → 224×224, Inception → 299×299   |
| **Training Input Shape** | The resized input used during training (via `target_size` in Keras).         | E.g., `target_size=(224, 224)`               |
| **Source Image Size**    | The raw size of the dataset images before resizing.                          | 500×500 or 256×256 depending on the test     |

- Most top-performing models were trained using 500×500 source images, with the exception of MobileNetV2, which showed similar performance with 256×256 sources.

- For MobileNetV2, using a 256×256 or 500×500 source image produced nearly identical results (F1-score between 0.58 and 0.59).  
  To simplify preprocessing and ensure consistency across models, we will adopt **500×500 as the standard raw image size** going forward.

- These results confirm the relevance of the selected models **under default training conditions** (no data augmentation or fine-tuning), and will guide the next steps focused on **data augmentation** and **fine-tuning**.
